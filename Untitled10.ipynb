```python

# Step 2: Import libraries and configure dataset paths

import os
import random
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
from torchvision import transforms, datasets
from torch.utils.data import DataLoader, random_split, Subset

from PIL import Image
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

# Dataset location
dataset_path = "/content/weather_dataset/dataset"

# Five selected weather classes
classes = ["cloudy", "foggy", "rainy", "shine", "sunrise"]

# Reproducibility: keep random operations consistent
random.seed(42)
np.random.seed(42)
torch.manual_seed(42)

# Select GPU if available; otherwise use CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("PyTorch version:", torch.__version__)
print("Torchvision version:", torchvision.__version__)
print("Device:", device)
print("Dataset path:", dataset_path)

print("\nChecking weather classes:")

for class_name in classes:
    folder_path = os.path.join(dataset_path, class_name)

    if os.path.isdir(folder_path):
        image_count = len([
            name for name in os.listdir(folder_path)
            if os.path.isfile(os.path.join(folder_path, name))
        ])
        print(f"{class_name}: folder found, {image_count} files")
    else:
        print(f"{class_name}: FOLDER NOT FOUND")
```

```python

# Step 3.1: Display sample images from all five classes

fig, axes = plt.subplots(5, 3, figsize=(10, 15))

for row, class_name in enumerate(classes):
    folder_path = os.path.join(dataset_path, class_name)

    image_files = sorted([
        name for name in os.listdir(folder_path)
        if os.path.isfile(os.path.join(folder_path, name))
    ])[:3]

    for col in range(3):
        ax = axes[row, col]

        if col < len(image_files):
            image_path = os.path.join(folder_path, image_files[col])

            with Image.open(image_path) as img:
                ax.imshow(img.convert("RGB"))

            ax.set_title(class_name.capitalize())
        else:
            ax.set_visible(False)

        ax.axis("off")

plt.tight_layout()
plt.show()
```

```python

# Step 3.2: Image preprocessing and normalization

image_size = 128

transform = transforms.Compose([
    transforms.Resize((image_size, image_size)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])

print("Image preprocessing pipeline created successfully!")
print("Target image size:", image_size, "x", image_size)
print("Image channels: RGB (3)")
print("Normalization: mean=0.5, standard deviation=0.5")
```

```python

# Step 3.3: Load images and split the dataset

full_dataset = datasets.ImageFolder(
    root=dataset_path,
    transform=transform
)

# Keep only our five selected weather classes.
# This excludes the alien_test folder.
selected_indices = [
    index
    for index, (_, label) in enumerate(full_dataset.samples)
    if full_dataset.classes[label] in classes
]

# Use a fixed random seed for a reproducible split
generator = torch.Generator().manual_seed(42)

selected_dataset = Subset(full_dataset, selected_indices)

train_size = int(0.70 * len(selected_dataset))
val_size = int(0.15 * len(selected_dataset))
test_size = len(selected_dataset) - train_size - val_size

train_dataset, val_dataset, test_dataset = random_split(
    selected_dataset,
    [train_size, val_size, test_size],
    generator=generator
)

print("Classes detected by ImageFolder:", full_dataset.classes)
print("Selected weather classes:", classes)
print("\nTotal selected images:", len(selected_dataset))
print("Training images:", len(train_dataset))
print("Validation images:", len(val_dataset))
print("Testing images:", len(test_dataset))
```

```python

# Step 4: Correct class labels and create DataLoaders

# Remap selected labels:
# cloudy 1 -> 0, foggy 2 -> 1, rainy 3 -> 2,
# shine 4 -> 3, sunrise 5 -> 4
full_dataset.target_transform = lambda label: label - 1

# Batch size: number of images processed together
batch_size = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False,
    num_workers=0
)

print("Class-to-index mapping:")
for index, class_name in enumerate(classes):
    print(f"{class_name}: {index}")

print("\nDataLoaders created successfully!")
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))

# Verify one batch
images, labels = next(iter(train_loader))

print("\nImage batch shape:", images.shape)
print("Label batch shape:", labels.shape)
print("Labels in this batch:", torch.unique(labels).tolist())
print("Image tensor range:",
      round(images.min().item(), 3),
      "to",
      round(images.max().item(), 3))
```

```python

# Step 5: Build the Convolutional Neural Network (CNN)

class WeatherCNN(nn.Module):

    def __init__(self, num_classes=5):
        super(WeatherCNN, self).__init__()

        # Feature extraction layers
        self.features = nn.Sequential(

            # Block 1: 3 input channels -> 32 feature maps
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),

            # Block 2: 32 -> 64 feature maps
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),

            # Block 3: 64 -> 128 feature maps
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2)
        )

        # Classification layers
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 16 * 16, 256),
            nn.ReLU(),
            nn.Dropout(p=0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


# Create the model and move it to the selected device
model = WeatherCNN(num_classes=len(classes)).to(device)

print(model)

# Count trainable parameters
total_params = sum(
    parameter.numel()
    for parameter in model.parameters()
    if parameter.requires_grad
)

print("\nNumber of trainable parameters:", total_params)
print("Model device:", device)
```

```python

# Step 6: Configure loss function and optimizer

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

num_epochs = 5

print("Loss function:", criterion)
print("Optimizer:", optimizer)
print("Learning rate:", 0.001)
print("Number of epochs:", num_epochs)
```

```python

# Step 6.2: Train the baseline CNN

train_losses = []
val_losses = []
train_accuracies = []
val_accuracies = []

for epoch in range(num_epochs):

    # Training phase
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        # Clear old gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = model(images)

        # Calculate loss
        loss = criterion(outputs, labels)

        # Backpropagation
        loss.backward()

        # Update model weights
        optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)
        total += labels.size(0)
        correct += (predictions == labels).sum().item()

    epoch_train_loss = running_loss / total
    epoch_train_accuracy = correct / total

    # Validation phase
    model.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_running_loss += loss.item() * images.size(0)

            predictions = torch.argmax(outputs, dim=1)
            val_total += labels.size(0)
            val_correct += (predictions == labels).sum().item()

    epoch_val_loss = val_running_loss / val_total
    epoch_val_accuracy = val_correct / val_total

    train_losses.append(epoch_train_loss)
    val_losses.append(epoch_val_loss)
    train_accuracies.append(epoch_train_accuracy)
    val_accuracies.append(epoch_val_accuracy)

    print(
        f"Epoch [{epoch + 1}/{num_epochs}] "
        f"| Train Loss: {epoch_train_loss:.4f} "
        f"| Train Accuracy: {epoch_train_accuracy * 100:.2f}% "
        f"| Val Loss: {epoch_val_loss:.4f} "
        f"| Val Accuracy: {epoch_val_accuracy * 100:.2f}%"
    )

print("\nBaseline training completed!")
```

```python

# Step 7: Plot training and validation curves

epochs_range = range(1, num_epochs + 1)

# Plot loss
plt.figure(figsize=(8, 5))
plt.plot(epochs_range, train_losses, marker="o", label="Training Loss")
plt.plot(epochs_range, val_losses, marker="o", label="Validation Loss")

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.xticks(list(epochs_range))
plt.legend()
plt.grid(True)
plt.show()

# Plot accuracy
plt.figure(figsize=(8, 5))
plt.plot(
    epochs_range,
    [accuracy * 100 for accuracy in train_accuracies],
    marker="o",
    label="Training Accuracy"
)
plt.plot(
    epochs_range,
    [accuracy * 100 for accuracy in val_accuracies],
    marker="o",
    label="Validation Accuracy"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.xticks(list(epochs_range))
plt.legend()
plt.grid(True)
plt.show()
```

```python

# Step 8: Evaluate the baseline CNN on the test dataset

model.eval()

y_true = []
y_pred = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)

        outputs = model(images)
        predictions = torch.argmax(outputs, dim=1)

        y_true.extend(labels.numpy().tolist())
        y_pred.extend(predictions.cpu().numpy().tolist())

# Calculate evaluation metrics
test_accuracy = accuracy_score(y_true, y_pred)

test_precision = precision_score(
    y_true, y_pred, average="macro", zero_division=0
)

test_recall = recall_score(
    y_true, y_pred, average="macro", zero_division=0
)

test_f1 = f1_score(
    y_true, y_pred, average="macro", zero_division=0
)

print("BASELINE CNN - TEST RESULTS")
print("-" * 35)
print(f"Accuracy:  {test_accuracy * 100:.2f}%")
print(f"Precision: {test_precision * 100:.2f}%")
print(f"Recall:    {test_recall * 100:.2f}%")
print(f"F1-score:  {test_f1 * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=list(range(len(classes))),
    target_names=classes,
    zero_division=0
))

# Confusion matrix
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=list(range(len(classes)))
)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=classes
)

fig, ax = plt.subplots(figsize=(8, 6))
display.plot(ax=ax, cmap="Blues", values_format="d")
plt.title("Baseline CNN - Confusion Matrix")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# Step 10.1: Create data augmentation pipeline

augmentation_transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=10),
    transforms.ColorJitter(
        brightness=0.15,
        contrast=0.15
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])

# Create a separate dataset with augmentation
augmented_full_dataset = datasets.ImageFolder(
    root=dataset_path,
    transform=augmentation_transform,
    target_transform=lambda label: label - 1
)

# Preserve the exact same split indices as the baseline
augmented_selected_dataset = Subset(
    augmented_full_dataset,
    selected_indices
)

augmented_train_dataset = Subset(
    augmented_selected_dataset,
    train_dataset.indices
)

print("Augmentation pipeline created.")
print("Training images:", len(augmented_train_dataset))
print("Validation images:", len(val_dataset))
print("Testing images:", len(test_dataset))
```

```python
# Step 10.2: Create the augmented training DataLoader

augmented_train_loader = DataLoader(
    augmented_train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=0
)

# Check one batch
images, labels = next(iter(augmented_train_loader))

print("Augmented training loader created successfully!")
print("Image batch shape:", images.shape)
print("Label batch shape:", labels.shape)
print("Labels in this batch:", torch.unique(labels).tolist())
print("Pixel value range:",
      round(images.min().item(), 2),
      "to",
      round(images.max().item(), 2))
```

```python
# Step 10.3: Train the improved CNN

import copy

# Create a separate improved model
improved_model = WeatherCNN(num_classes=len(classes)).to(device)

# Loss function
improved_criterion = nn.CrossEntropyLoss()

# Adam optimizer with L2 regularization
improved_optimizer = optim.Adam(
    improved_model.parameters(),
    lr=0.0005,
    weight_decay=0.0001
)

# Training settings
improved_epochs = 5

improved_train_losses = []
improved_val_losses = []
improved_train_accuracies = []
improved_val_accuracies = []

best_val_loss = float("inf")
best_model_state = None

for epoch in range(improved_epochs):

    # -------------------------
    # Training phase
    # -------------------------
    improved_model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in augmented_train_loader:

        images = images.to(device)
        labels = labels.to(device)

        improved_optimizer.zero_grad()

        outputs = improved_model(images)
        loss = improved_criterion(outputs, labels)

        loss.backward()
        improved_optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    train_loss = running_loss / total
    train_accuracy = 100 * correct / total

    improved_train_losses.append(train_loss)
    improved_train_accuracies.append(train_accuracy)

    # -------------------------
    # Validation phase
    # -------------------------
    improved_model.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = improved_model(images)
            loss = improved_criterion(outputs, labels)

            val_running_loss += loss.item() * images.size(0)

            predictions = torch.argmax(outputs, dim=1)
            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

    val_loss = val_running_loss / val_total
    val_accuracy = 100 * val_correct / val_total

    improved_val_losses.append(val_loss)
    improved_val_accuracies.append(val_accuracy)

    # Save the model with the lowest validation loss
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_model_state = copy.deepcopy(improved_model.state_dict())

    print(
        f"Epoch [{epoch + 1}/{improved_epochs}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Accuracy: {train_accuracy:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Accuracy: {val_accuracy:.2f}%"
    )

# Restore the best validation model
improved_model.load_state_dict(best_model_state)

print("\nImproved CNN training completed!")
print(f"Best validation loss: {best_val_loss:.4f}")
```

```python
# Step 10.4: Evaluate the improved CNN

improved_model.eval()

improved_y_true = []
improved_y_pred = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)

        outputs = improved_model(images)
        predictions = torch.argmax(outputs, dim=1)

        improved_y_true.extend(labels.numpy().tolist())
        improved_y_pred.extend(predictions.cpu().numpy().tolist())

# Calculate evaluation metrics
improved_accuracy = accuracy_score(
    improved_y_true, improved_y_pred
)

improved_precision = precision_score(
    improved_y_true,
    improved_y_pred,
    average="macro",
    zero_division=0
)

improved_recall = recall_score(
    improved_y_true,
    improved_y_pred,
    average="macro",
    zero_division=0
)

improved_f1 = f1_score(
    improved_y_true,
    improved_y_pred,
    average="macro",
    zero_division=0
)

print("IMPROVED CNN TEST RESULTS")
print("-" * 35)
print(f"Accuracy:  {improved_accuracy * 100:.2f}%")
print(f"Precision: {improved_precision * 100:.2f}%")
print(f"Recall:    {improved_recall * 100:.2f}%")
print(f"F1-score:  {improved_f1 * 100:.2f}%")

print("\nClassification Report:")
print(
    classification_report(
        improved_y_true,
        improved_y_pred,
        labels=list(range(len(classes))),
        target_names=classes,
        zero_division=0
    )
)

# Plot confusion matrix
improved_cm = confusion_matrix(
    improved_y_true,
    improved_y_pred,
    labels=list(range(len(classes)))
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=improved_cm,
    display_labels=classes
)

disp.plot(cmap="Blues", xticks_rotation=45)
plt.title("Improved CNN - Test Confusion Matrix")
plt.tight_layout()
plt.show()
```
